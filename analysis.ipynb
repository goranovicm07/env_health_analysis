## <span style="color: rgb(39, 55, 77); background: rgb(221, 230, 237)">Part 2: Situation Background</span>

For decades, scientists, ecologists and other professional researchers have pondered the impact our environment has on overall health and well-being. In recent years, an increased awareness in environmental changes, environmental contaminants, urban development, infrastructure, and social factors has called for more comprehensive approaches to understanding the relationship between the envrionment and human health. For this reason, the United States Envrionmental Protection Agency (U.S. EPA) developed the Environmental Quality Index (EQI) in the early 2000s.

The EQI is an assessment tool that evaluates environmental quality and potential impacts on human health across geographic regions, typically counties. The environmental quality is divided by domains: air quality, water quality, land quality, built environment, and sociodemographic environment. To determine the EQI score, air pollutants, water pollutants, land pollutants, factors related to urban development and infrastructure, and social factors are measured, normalized, and aggregated into a composite score by experts. 

Per the EPA, "Since the release of the EQI 2000-2005, researchers have published multiple studies examining the relationship between overall environmental quality and health outcomes, including preterm birth, mortality, cancer incidence, asthma prevalence, physical inactivity and obesity, infant mortality, and pediatric multiple sclerosis." While these are excellent studies that found significant results, my research specifically seeks to address the relationship between environmental quality, life expectancy, and cancer incidence rates in Wisconsin counties. Furthermore, my results will not directly use the EQI values. Instead, the data that is used to determine the EQI will be used. That way, the specific pollutants and factors that affect life expectancy and cancer rates can be examined.

Hence, my research seeks to address three key questions.

### Statistical Tests

**Set Up For Quantitative (Statistical) Analysis**

A quantitative analysis will now be performed. Before starting, several steps will be performed to ensure the statistical tests run smoothly.

The first step is to simply re-define the targets and candidate columns, which exclude the leakage / ID columns.

```python
targets = ["Total", "Crude Rate"]
candidate_cols = [
    c
    for c in data.columns
    if c not in features_to_drop and pd.api.types.is_numeric_dtype(data[c])
]

print(len(candidate_cols))
```

The second step is to separate the constant columns (`w_deph_ln` and `w_alpha_ln`) from the active numeric predictors. This is because a Pearson Correlation test will be performed, and by excluding the constant columns, a `ConstantInputWarning` and `NaN` correlations will be avoided.

```python
const_cols = [c for c in candidate_cols if data[c].nunique(dropna=True) <= 1]
active_cols = [c for c in candidate_cols if c not in const_cols]

print(f"Removed {len(const_cols)} constant columns:", const_cols)
print(f"Active numeric predictors remaining: {len(active_cols)}\n")
```

The third step is to flag heavily zero-inflated or duplicate features. From the above analysis, it was determined that many features are 80%-97% zeros or nearly correlated. Pearson's Correlation test can be driven by just a handful of non-zero counties (or a single outlier county), so these will be flagged.

```python
sparsity_flags = []
for col in active_cols:
  s = data[col]
  pct_zeros = (s == 0).mean() * 100
  min_val = s.min()
  pct_at_min = (s == min_val).mean() * 100
  if pct_zeros >= 80.0 or pct_at_min >= 50.0:
    sparsity_flags.append({
        "feature": col,
        "n_unique": s.nunique(dropna=True),
        "pct_zeros": round(pct_zeros, 1),
        "min_val": round(min_val, 6),
        "pct_at_min": round(pct_at_min, 1),
    })

sparsity_df = pd.DataFrame(sparsity_flags)
print(f"Flagged {len(sparsity_df)} zero-inflated or floor-clumped features:")
if not sparsity_df.empty:
  print(sparsity_df.to_string(index=False))
print()
```

```python
corr_abs = data[active_cols].corr().abs()
duplicate_pairs = []
for i in range(len(active_cols)):
  for j in range(i + 1, len(active_cols)):
    r_abs = corr_abs.iloc[i, j]
    if r_abs >= 0.999:
      r_signed = data[active_cols[i]].corr(data[active_cols[j]])
      duplicate_pairs.append({
          "feature_1": active_cols[i],
          "feature_2": active_cols[j],
          "r": round(r_signed, 6),
      })
```

```python
dup_df = pd.DataFrame(duplicate_pairs)
print(f"Found {len(dup_df)} near-identical predictor pairs (|r| >= 0.999):")
if not dup_df.empty:
  print(dup_df.head(25).to_string(index=False))
```

There are 130 pairs with $|r|>0.999$. This explains a lot of the redundancy in the over 200 features. It appears that water (`w_*_ln`) and air (`a_*_ln`) cluster at $r \approx 1.0$

I do not want to delete all duplicates blindly, while this could help achieve a non-redundant feature set for modeling. Instead, I want to group these feature into connected clusters so I can see how many distinct feature groups actually exist among my active predictors. 

```python
corr_abs = data[active_cols].corr().abs()
visited = set()
clusters = []

for col in active_cols:
    if col in visited:
        continue
    stack = [col]
    comp = []
    visited.add(col)
    while stack:
        curr = stack.pop()
        comp.append(curr)
        for nxt in active_cols:
            if nxt not in visited and corr_abs.loc[curr, nxt] >= 0.999:
                visited.add(nxt)
                stack.append(nxt)
    clusters.append(comp)

multi_clusters = [c for c in clusters if len(c) > 1]
print(
    f"Total active features: {len(active_cols)} -> Collapsed into"
    f" {len(clusters)} distinct groups ({len(multi_clusters)} multi-feature"
    " clusters)."
)
for idx, comp in enumerate(multi_clusters[:10], start=1):
    print(f"Group {idx} ({len(comp)} cols): {comp}\n")
```

These are the exact groups of variables that are nearly collinear across WI counties. The remaining 191 features (+ one from each column 191 + 5 = 196) are distinct or non-redundant.

**Bivariate Correlation Testing and Bonferroni Correction**

To evaluate how each active predictor relates to both county-level outcomes (`Total` life expectancy and cancer `Crude Rate`), the next cell will compute both Pearson's Correlation (measuring linear relationships) and Spearman's Rank Correlation (measuring monotonic rank-order relationships, which is helpful given the skewness, zeros, and detection-limit floors observed during EDA).

Because running hundreds of hypothesis tests increases the family-wise error rate (the chance of false positives), a Bonferroni correction ($\alpha_{\text{adj}} = 0.05 / m$, where $m$ is the number of tested features in `analysis_cols`) is applied to set a stricter significance threshold before storing all results in `corr_results_df`.

```python
bonferroni_alpha = 0.05 / len(analysis_cols)

all_results = []

for target in targets:
  y = data[target]

  for col in analysis_cols:
    x = data[col]

    r_pearson, p_pearson = stats.pearsonr(x, y)
    rho_spearman, p_spearman = stats.spearmanr(x, y)

    all_results.append({
        "target": target,
        "feature": col,
        "pearson_r": r_pearson,
        "pearson_p": p_pearson,
        "pearson_bonf_sig": p_pearson < bonferroni_alpha,
        "spearman_rho": rho_spearman,
        "spearman_p": p_spearman,
        "spearman_bonf_sig": p_spearman < bonferroni_alpha,
    })

corr_results_df = pd.DataFrame(all_results)

print(f"Tested {len(analysis_cols)} features per target")
print(f"Bonferroni-adjusted alpha (0.05 / {len(analysis_cols)}): {bonferroni_alpha:.6e}\n")
```

**Ranking Top Correlates and Extracting Bonferroni-Significant Features**

With all correlation coefficients, p-values, and Bonferroni flags stored in `corr_results_df`, the following cell will filter and sort the results separately for `Total` and `Crude Rate` to highlight the strongest associations.

For each target, the top 10 features ranked by Pearson p-value will be displayed, alongside their Spearman rank statistics. This will enable me to compare linear and rank-based consist comparison. It will also extact the lists of features that pass the Bonferroni-adjusted threshold for both tests.

```python
for target in targets:
    sub = corr_results_df[corr_results_df["target"] == target].sort_values(
        "pearson_p", ascending=True
    )

    print(f"=== Target: {target} ===")
    print("\nTop 10 features by Pearson p-value:\n")
    print(
        sub[[
            "feature",
            "pearson_r",
            "pearson_p",
            "pearson_bonf_sig",
            "spearman_rho",
            "spearman_p",
            "spearman_bonf_sig",
        ]]
        .head(10)
        .to_string(index=False)
    )

    bonf_pearson_feats = sub.loc[sub["pearson_bonf_sig"], "feature"].tolist()
    bonf_spearman_feats = sub.loc[sub["spearman_bonf_sig"], "feature"].tolist()

    print(
        f"\nBonferroni-significant features ({target}) - Pearson:\n"
        f"\n({len(bonf_pearson_feats)}): {bonf_pearson_feats}"
    )
    print(
        f"\nBonferroni-significant features ({target}) - Spearman\n"
        f"\n({len(bonf_spearman_feats)}): {bonf_spearman_feats}\n"
    )
```

**Insights**

There are strong socioeconomic and demographic signals. 

For `Total` life expectancy, features like `pct_unemp`, `pct_pers_lt_pov`, `med_hh_inc`, `pct_hs_more`, `pct_vac_units`, and `mg_ln_ave` all pass the strict Bonferroni threshold on both Pearson and Spearman tests.

For cancer `Crude Rate`, features like `pct_vac_units`, `med_rooms`, `pct_rent_occ`, `pct_mt_10units_log`, `a_ph3_ln`, `a_diesel_ln` are among the strongest correlates. It looks like housing / vacancy / multi-unit rates often track age structure, urban vs. rural county profiles, and socioeconomic conditions.

**Insights Continued**

There are many Bonferroni hits - there are 46 (Pearson) and 59 (Spearman) significant features for `Crude Rate` across only 72 counties. This likely signals they are clustered or confounded. It is recommended to not throw all of these into an unregularized regression model.

### Linear Regression (Unregularized and Regularized)

**Unregularized Linear Regression**

I would first like to demonstrate the $p > n$ overfitting problem (and multicollinearity). As mentioned, with 72 counties and 196 features, unregularized ordinary least squares (or linear regression) has more predictors than observations ($p > n$). Training $R^2$ can look artifically high while out-of-sample cross-validated $R^2$ collapses (often turning negative) and coefficients swing wildly.

```python
print(len(rep_features))
print(rep_features)
```

```python
targets = ["Total", "Crude Rate"]
cv = KFold(n_splits=5, shuffle=True, random_state=24)

for target in targets:
    X = data[rep_features].copy()
    y = data[target].copy()

    ols_pipe = Pipeline(
        [("scaler", StandardScaler()), ("model", LinearRegression())]
    )

    ols_pipe.fit(X, y)
    y_in_pred = ols_pipe.predict(X)
    y_cv_pred = cross_val_predict(ols_pipe, X, y, cv=cv)

    coefs = pd.Series(
        ols_pipe.named_steps["model"].coef_, index=rep_features
    ).sort_values(key=abs, ascending=False)

    print(f"=== Unregularized OLS Baseline: {target} (n=72, p={X.shape[1]}) ===")
    print(f"In-sample R2:  {r2_score(y, y_in_pred):.4f}")
    print(
        f"5-fold CV R2:  {r2_score(y, y_cv_pred):.4f} | CV RMSE:"
        f" {np.sqrt(mean_squared_error(y, y_cv_pred)):.4f} | CV MAE:"
        f" {mean_absolute_error(y, y_cv_pred):.4f}"
    )
    print("Largest 8 |coefficient| magnitudes (standardized):")
    print(coefs.head(8).round(3).to_string(), "\n")
```

**Insights**

It appears unregularized OLS overfits severely, especially on `Crude Rate`. 

On both targets, unregularized Linear Regression gets an in-sample $R^2$ of 1.0000 because $p=196$ easily memorizes 72 counties. Out-of-sample, 5-fold CV drops to 0.2781 for `Total` and collapses to -0.3965 (which is worse than predicting the mean) for `Crude Rate`, with RMSE jumping to 139.89. For the `Crude Rate`, OLS standardized coefficients reach ~16.7 to ~24.4 in opposite directions (`pct_vac_units` at +24.44 and `pct_rent_occ` at -21.42), which looks like classic collinearity.

**Regularized Models (Ridge, Lasso, Elastic Net)**

Regularization shrinks coefficients toward zero (Ridge uses L2 penalty; Lasso uses L1 penalty and does sparse selection; ElasticNet combines L1 and L2, which often behaves better when correlated predictors remain). I will fit inside Pipeline + KFold to keep scaling and penalty tuning (*CV) properly cross-validated.

```python
alphas_ridge = np.logspace(-1, 4, 60)
alphas_lasso = np.logspace(-3, 1.5, 60)
l1_ratios = [0.2, 0.5, 0.8, 0.95]

reg_summary = []
top_lasso_by_target = {}

for target in targets:
    X = data[rep_features].copy()
    y = data[target].copy()

    models = {
        "RidgeCV": Pipeline([
            ("scaler", StandardScaler()),
            ("model", RidgeCV(alphas=alphas_ridge, cv=cv)),
        ]),
        "LassoCV": Pipeline([
            ("scaler", StandardScaler()),
            ("model", LassoCV(alphas=alphas_lasso, cv=cv, max_iter=20000, random_state=24)),
        ]),
        "ElasticNetCV": Pipeline([
            ("scaler", StandardScaler()),
            (
                "model",
                ElasticNetCV(
                    alphas=alphas_lasso,
                    l1_ratio=l1_ratios,
                    cv=cv,
                    max_iter=20000,
                    random_state=24,
                ),
            ),
        ]),
    }

    print(f"=== Regularized Models on rep_features (p={X.shape[1]}): {target} ===")
    for name, pipe in models.items():
        pipe.fit(X, y)
        y_cv = cross_val_predict(pipe, X, y, cv=cv)
        rmse = np.sqrt(mean_squared_error(y, y_cv))
        mae = mean_absolute_error(y, y_cv)
        r2_cv = r2_score(y, y_cv)

        est = pipe.named_steps["model"]
        n_nonzero = int(np.sum(np.abs(est.coef_) > 1e-6))

        reg_summary.append({
            "target": target,
            "model": name,
            "cv_r2": round(r2_cv, 4),
            "cv_rmse": round(rmse, 4),
            "cv_mae": round(mae, 4),
            "nonzero_coefs": n_nonzero,
            "best_alpha": round(float(est.alpha_), 5),
        })

        if name == "LassoCV":
            coef_s = pd.Series(est.coef_, index=rep_features)
            nonzero_s = coef_s[coef_s.abs() > 1e-6].reindex(
                coef_s[coef_s.abs() > 1e-6].abs().sort_values(ascending=False).index
            )
            top_lasso_by_target[target] = nonzero_s

    res_sub = pd.DataFrame([r for r in reg_summary if r["target"] == target])
    print(res_sub.drop(columns=["target"]).to_string(index=False))
    print("Top non-zero LassoCV features:")
    print(top_lasso_by_target[target].head(12).round(4).to_string(), "\n")
```

**Insights**

It looks like regularization dramatically improves out-of-sample generalization. 

For `Total` life expectancy, Elastic Net on all 196 features reaches CV $R^2$ = 0.6422, cutting CV RMSE from 1.2269 down to 0.8638.

The top sparse drivers in Lasso combined socioeconomic / housing structure (`pct_unemp`, `med_rooms`, `pct_hs_more`, `pct_vac_units`) with specific environmental variables (`w_ba_ln`, `a_c3h3n_ln`, `a_isophorone_ln`, `insecticides_ln`, `w_deha_ln`).

For `Crude Rate`, Lasso reaches CV $R^2$ = 0.6382 (RMSE 71.2057) using just 15 non-zero features, with `pct_vac_units` (62.18), `rate_trans_env_log` (22.88), `a_ph3_ln` (-22.24), and `pct_rent_occ` (-16.51) leading.

**Controlled and Domain-Targeted Models for RQ1 and RQ2**

Now, I would like to directly test RQ1 (contaminants vs. `Total` life expectancy) and RQ2 (contaminants vs. cancer `Crude Rate`) by comparing:

- Confounders only (`confounder_cols`): Socioeconomic / demographic baseline
- Confounders + All Environmental Representatives (`confounder_cols` + `env_rep_features`): Tests whether environmental variables improve out of sample CV performance over demographics alone
- Confounders + Domain Subsets: air (`a_*`), water (`w_*`), and agricultural / land-use (`pct_*` / farming / acreage) subsets added one at a time on top of `confounder_cols` so I can see which specific exposure domain adds explanatory power.

```python
alphas_ridge = np.logspace(-1, 4, 60)
alphas_lasso = np.logspace(-1.5, 2.0, 60)
l1_ratios = [0.2, 0.5, 0.8, 0.95]

air_cols = [c for c in env_rep_features if c.startswith("a_")]
water_cols = [c for c in env_rep_features if c.startswith("w_")]
ag_keywords = ("acre", "farm", "manure", "herbicide", "insecticide", "fertiliz", "crop")
ag_cols = [
    c for c in env_rep_features if any(k in c.lower() for k in ag_keywords)
]

feature_sets = {
    "1_Confounders_Only": confounder_cols,
    "2_Confounders_+_All_Env": confounder_cols + env_rep_features,
    "3_Confounders_+_Air": confounder_cols + air_cols,
    "4_Confounders_+_Water": confounder_cols + water_cols,
    "5_Confounders_+_Ag_Land": confounder_cols + ag_cols,
}

print("Feature set sizes:", {k: len(v) for k, v in feature_sets.items()}, "\n")

controlled_rows = []

for target in targets:
    y = data[target].copy()

    for set_name, cols in feature_sets.items():
        if len(cols) == 0:
            continue
        X = data[cols].copy()

        candidates = {
            "RidgeCV": Pipeline([
                ("scaler", StandardScaler()),
                ("model", RidgeCV(alphas=alphas_ridge, cv=cv)),
            ]),
            "ElasticNetCV": Pipeline([
                ("scaler", StandardScaler()),
                (
                    "model",
                    ElasticNetCV(
                        alphas=alphas_lasso,
                        l1_ratio=l1_ratios,
                        cv=cv,
                        max_iter=50000,
                        tol=1e-3,
                        random_state=24,
                    ),
                ),
            ]),
        }

        for m_name, pipe in candidates.items():
            y_cv = cross_val_predict(pipe, X, y, cv=cv)
            controlled_rows.append({
                "target": target,
                "feature_set": set_name,
                "n_features": len(cols),
                "model": m_name,
                "cv_r2": round(r2_score(y, y_cv), 4),
                "cv_rmse": round(np.sqrt(mean_squared_error(y, y_cv)), 4),
                "cv_mae": round(mean_absolute_error(y, y_cv), 4),
            })

controlled_df = pd.DataFrame(controlled_rows)

for target in targets:
    rq_label = "RQ1 (Life Expectancy)" if target == "Total" else "RQ2 (Cancer Crude Rate)"
    print(f"=== {target} — {rq_label} ===")
    sub = controlled_df[controlled_df["target"] == target].drop(columns=["target"])
    print(sub.to_string(index=False), "\n")
```

**Insights**

It looks like confounders do much of the heavy lifting, especially for `Crude Rate`.

For `Crude Rate`, confounders only (12 features) with Ridge achieves the highest CV $R^2$, beating confounders and environmental contaminants and every domain subset. Adding 184 environmental features actually hurts out-of-sample CV accuracy on `Crude Rate` likely because extra noise predictors dilute regularization across 72 rows.

For `Total`, confounders only get an $R^2$ = 0.4095, while confounders + environmental features and confounders + water with Elastic Net jump substantially ($R^2$ >= 0.5833), suggesting environmental / water features add incremental predictive value for life expectancy beyond those 12 controls.

It was interesting to notice how confounders + water on `Total` had Ridge drop to -0.0605 while Elastic Net hit 0.6386. When 51 water / confounder features include noisy or floor-clumped columns, L2-only shrinkage at a fixed grid can over-smooth while L1 / Elastic net feature selection isolates useful signals.

### Random Forest

The Ridge and Elastic Net models already perform strongly and transparently on this dataset. However, I would like to also try Random Forest as a secondary non-linear benchmark. Some environmental exposures, housing / demographic relationships, etc. can be non-linear or threshold-based. Tree ensembles can naturally capture these non-linearities and feature interactions without requring manual interaction terms. Also, comparing a linear / regularized model against a tree model gives a nice check on whether relationships in my data are mostly linear or non-linear. That being said, there are disadvantages given the small sample size and high dimensionality.

```python
targets = ["Total", "Crude Rate"]
cv = KFold(n_splits=5, shuffle=True, random_state=24)

air_cols = [c for c in env_rep_features if c.startswith("a_")]
water_cols = [c for c in env_rep_features if c.startswith("w_")]
ag_keywords = (
    "acre",
    "farm",
    "manure",
    "herbicide",
    "insecticide",
    "fertiliz",
    "crop",
)
ag_cols = [
    c for c in env_rep_features if any(k in c.lower() for k in ag_keywords)
]

feature_sets = {
    "1_Confounders_Only": confounder_cols,
    "2_Confounders_+_All_Env": confounder_cols + env_rep_features,
    "3_Confounders_+_Air": confounder_cols + air_cols,
    "4_Confounders_+_Water": confounder_cols + water_cols,
    "5_Confounders_+_Ag_Land": confounder_cols + ag_cols,
}
```

```python
rf_model = RandomForestRegressor(
    n_estimators=500,
    max_depth=5,
    min_samples_leaf=5,
    max_features="sqrt",
    n_jobs=-1,
    random_state=24,
)

rf_rows = []
top_rf_importances = {}

for target in targets:
    y = data[target].copy()
    target_imps = {}

    for set_name, cols in feature_sets.items():
        if len(cols) == 0:
            continue
        X = data[cols].copy()

        y_cv = cross_val_predict(rf_model, X, y, cv=cv)
        rf_rows.append({
            "target": target,
            "feature_set": set_name,
            "n_features": len(cols),
            "model": "RandomForest (reg)",
            "cv_r2": round(r2_score(y, y_cv), 4),
            "cv_rmse": round(np.sqrt(mean_squared_error(y, y_cv)), 4),
            "cv_mae": round(mean_absolute_error(y, y_cv), 4),
        })

        rf_model.fit(X, y)
        imp_s = pd.Series(rf_model.feature_importances_, index=cols).sort_values(
            ascending=False
        )
        target_imps[set_name] = imp_s

    top_rf_importances[target] = target_imps

rf_df = pd.DataFrame(rf_rows)

for target in targets:
    rq_label = (
        "RQ1 (Life Expectancy)"
        if target == "Total"
        else "RQ2 (Cancer Crude Rate)"
    )
    print(f"=== {target} — {rq_label} (Regularized Random Forest) ===")
    sub = rf_df[rf_df["target"] == target].drop(columns=["target"])
    print(sub.to_string(index=False))

    if "2_Confounders_+_All_Env" in top_rf_importances[target]:
        print("\nTop 8 importances (2_Confounders_+_All_Env):")
        print(
            top_rf_importances[target]["2_Confounders_+_All_Env"]
            .head(8)
            .round(4)
            .to_string()
        )
    print("\n")

if "controlled_df" in globals():
    combined_df = (
        pd.concat([controlled_df, rf_df], ignore_index=True)
        .sort_values(["target", "feature_set", "cv_r2"], ascending=[True, True, False])
    )
    print("=== Combined Comparison (RidgeCV, ElasticNetCV, RandomForest) ===")
    print(combined_df.to_string(index=False))
```

**Insights**

The linear regularized models outperform Random Forest.

On `Total`, Elastic Net reaches CV $R^2$ = 0.6473 on confounders + all environmental features and 0.6386 on confounders + water features, whereas Random Forest tops out at 0.3630 (confounders only) and drops to 0.2168 when all 196 features are included.

On `Crude Rate`, Ridge on confounders only leads at CV $R^2$ = 0.6938 and Elastic Net reaches 0.6456 with air features and 0.6277 with water features - while Random Forest peaks at 0.5057 on confounders only and drops into the 0.39 - 0.47 range as more predictors are added.

**Insights Continued**

Even though the Random Forest model has lower out of sample $R^2$, its top impurity importances echo many of same signals seen in bivariate correlations and Lasso. 

For `Total`, housing / socioeconomic variables recur: `med_rooms`, `pct_unemp`, `pct_hs_more`, `pct_vac_units`, and `k_ln_ave`.

For `Crude Rate`, housing / socioeconomic also recur: (`pct_vac_units`, `pct_rent_occ`).

**Final Takeaways**

For `Total` life expectancy, confounders alone get ~0.41-0.42 CV $R^2$, and adding environmental predictors (Elastic Net on all environmental features or water + confounders) lifts CV $R^2$ to ~0.64-0.65, pointing to meaningful combined socioeconomic + environmental predictive structure.

For Cancer `Crude Rate`, the 12 feature confounder / demographic set alone hits 0.6938 CV $R^2$ with Ridge, showing that county demographic / housing structure (which proxies age distribution and urban / rural context) already explains most of the predictable variance in unadjusted crude cancer rates.

## <span style="color: rgb(39, 55, 77); background: rgb(221, 230, 237)">Part 13: Conclusions</span>

### RQ1: Environmental Contaminants and County Life Expectancy (`Total`)

**Primary Finding**

Environmental exposures show measurable associations with county-level life expectancy, and adding environmental predictors improves out-of-sample predictive accuracy beyond socioeconomic controls alone (though socioeconomic context remains foundational).

**Statistical and Modeling Evidence**

Bivariate screening with Bonferroni correlation ($\alpha_{adj}$ = 2.29 x 10^{-4}$) identified strong socioeconomic gradients alongisde significant environmental correlations.

In controlled cross-validated modeling ($N = 72$), socioeconomic confounders alone achived a 5-fold CV $R^2$ of 0.4158 (Elastic Net, RMSE = 1.1038 years). Expanding the feature set to include all environmenal representatives lifted 5-fold CV $R^2$ to 0.6473 (RMSE = 0.8577), with water-focused predictors accounting for much of that lift (CV $R^2$ = 0.6386).

Sparse feature selection (Lasso / Elastic Net) consistently paired socioeconomic and housing indicators (`pct_unemp`, `med_rooms`, `pct_hs_more`, `pct_vac_units`) with specific chemical and agricultural variables (such as `w_ba_ln`, `a_c3h3n_ln`, `a_isophorone_ln`, `insecticides_ln`, and `w_deha_ln`).

**Outliers**

County life expectancy averaged 79.43 years across Wisconsin, with below-average outcomes spanning both rural northern counties (Ashland, Sawyer, Washburn, Forest) and urban centers (Milwaukee, Kenosha). Menominee County emerged as a pronounced low outlier (72.5 years), underscoring severe localized socioeconomic and health disparities.

### RQ2: Environmental Contaminants and County Cancer Rates (`Crude Rate`)

**Primary Finding**

While numerous environmental variables exhibit statistically significant raw correlations with county cancer `Crude Rate`, demographic and housing confounders explain virtually all predictable out-of-sample variance, and many raw environmental correlations invert direction due to spatial and age-structure confounding.

**Statistical and Modeling Evidence**

In controlled modeling, the 12-feature socioeconomic and housing confounder baseline achieved the highest overall generalization performance using Ridge (CV $R^2$ = 0.6938, RMSE = 65.51). Adding all 184 environmental predictors, reduced out-of-sample CV $R^2$ to 0.5533 / 0.5731 (Ridge / Elastic Net), indicating that high-dimensional environmental features introduce noise rather than incremental pwoer for unadjusted cancer rates.

Scatterplots and bivariate correlations revealed counterintuitive inverse slopes for several industrial and agricultural pollutants (`a_diesel_ln`, `a_hcl_ln`, `a_ph3_ln`, `pct_manure_acres_ln`, `herbicides_ln`), where higher exposure levels coincided with *lower* crude cancer rates.

**Outliers**

High `Crude Rate` counties (state mean $\approx$ 595.17) are heavily concentrated in northern and central Wisconsin - led by Vilas ($> 1,000$), Iron, Adams, and Oneida (>800) - which have high shares and seasonal / vacant housing (`pct_vac_units`) and older retirement populations, alongside lower heavy industrial / agricultural intensity.

Conversely, Menominee County acts as a high-leverage bivariate outlier: despite having Wisconsin's highest unemployment rate (>16%), its crude cancer rate (~481) sits below the state mean - a pattern consist with a younger population age structure and competing mortality from shorter life expectancy.

### RQ3: Relationship Between Life Expectancy (`Total`) and Cancer Rates (`Crude Rate`) Based on Environmental Factors

**Primary Finding**

There is a moderate negative raw association, largely driven by shared confounders and geography. However, they is no link through a single shared environmental mechanism. Once socioeconomic / demographic confounders and environmental predictors are jointly accounted for, the residual correlation between the two outcomes drops close to zero, and shared predictors often act in opposing directions or reflect underlying demographic / geographic structure.

**Evidence**

Directly across Wisconsin's 72 counties, `Total` life expectancy and cancer `Crude Rate` correlate negatively: Pearson $r$ = -0.3148 ($p$ = 0.0071) and Spearman $\rho$ = -0.4406 ($p$ = 0.000107). This means counties with lower life expectancy tend to have higher adjusted crude cancer rates (and vice versa), with rank-order association stronger than linear association.

However, environmental features alone do not explain away the link. Combined confounders + environmental features do. Controlling for all environmental representives alone leaves the residual correlation virtually unchanged at $r$=-0.3146 ($p$=0.0071) - so environmental variables by themselves do not remove the inverse relationship between `Total` and `Crude Rate`. Controlling for confounders only (12) weakens it to $r$ = -0.2097 ($p$ = 0.0770), and controlling for confounders + environmental features (196) drops it near zero at $r$ = -0.0703 ($p$ = 0.5572). This suggests the apparent relationship between `Total` and `Crude Rate` is not a simple single-contaminant effect. Rather, it arises from joint socioeconomic, demographic / housing, and environmental structure across counties.

**Final Takeaway**

It was observed that there are opposite signs on shared predictors ($n = 18$ overlapping Elastic Net features). `pct_vac_units` (-0.2063 on `Total` vs. 29.6712 on `Crude Rate`) and `a_c3h3n_ln` (-0.3113 on `Total` vs. +4.6496 on `Crude Rate`). These move in opposite directions across the two outcomes. 

Meanwhile, several socioeconomic / housing features (`med_hh_inc`, `pct_hs_more`, `med_rooms`) have positive coefficients for `Total` and negative coefficients for `Crude Rate`, while several chemical / agricultural variables (`a_biphenyl_ln`, `a_me2_phthalate_ln`, `a_mehydrazine_ln`, `pct_manure_acres_ln`) show negative coefficients for `Crude Rate`, which once again reflects unadjusted age structure and urban / rural / northern-county differences rather than protective contaminant effects.

## <span style="color: rgb(39, 55, 77); background: rgb(221, 230, 237)">Part 14: Limitations and Future Work</span>

**Unadjusted Crude Cancer Rates (Age Confounding):** Because `Crude Rate` is not age-adjusted, counties with older populations have higher crude rates and younger counties have lower crude rates independent of environmental risk. Future work should analyze age-adjusted cancer incidence / mortality rates and include explicit county age-distribution controls.

**Ecological County-Level Aggregation ($N = 72$):** County-level metrics cannot capture individual exposure levels (such as private well contamination or census-tract disparities), and small sample size ($N = 72$) makes estimates sensitive to high-leverage counties like Menominee and Vilas.

**Multicollinearity & Proxy Interpretation:** Even after deduplicating $|r| > 0.999$ clusters down to 196 representative features, correlated chemical, agricultural, and socioeconomic variables mean individual non-zero coefficients in Lasso or ElasticNet often represent broader correlated regimes rather than isolated causal effects.

**Observational, Cross-Sectional Design:** This project evaluates statistical association and out-of-sample predictive performance rather than causal effects, without longitudinal exposure lags needed to account for multi-year cancer latency.

The third step is to build a tidy county-level export table titled `wi_counties_tableau.csv`.

```python
county_df = pd.DataFrame(index=data.index)
for col in id_cols:
    county_df[col] = data[col]

for target in ["Total", "Crude Rate"]:
    if target in data.columns:
        county_df[target] = data[target]
        county_df[f"{target}_vs_mean"] = data[target] - data[target].mean()

for col in available_predictors:
    county_df[col] = data[col]
```

**Step 3 Continued**

Now, out-of-fold predictions and residuals from primary models will be added.

```python
X_all = data[confounder_cols + env_rep_features].copy()
X_conf = data[confounder_cols].copy()

if "Total" in data.columns:
    en_total_pipe = Pipeline([
        ("scaler", StandardScaler()),
        (
            "model",
            ElasticNetCV(
                alphas=alphas_lasso,
                l1_ratio=l1_ratios,
                cv=cv,
                max_iter=50000,
                tol=1e-3,
                random_state=24,
            ),
        ),
    ])
    oof_total_en = cross_val_predict(en_total_pipe, X_all, data["Total"], cv=cv)
    county_df["oof_pred_total_elasticnet_all_env"] = oof_total_en
    county_df["oof_resid_total_elasticnet_all_env"] = data["Total"] - oof_total_en

if "Crude Rate" in data.columns:
    ridge_cancer_pipe = Pipeline([
        ("scaler", StandardScaler()),
        ("model", RidgeCV(alphas=alphas_ridge, cv=cv)),
    ])
    oof_cancer_ridge = cross_val_predict(
        ridge_cancer_pipe, X_conf, data["Crude Rate"], cv=cv
    )
    county_df["oof_pred_cancer_ridge_confounders"] = oof_cancer_ridge
    county_df["oof_resid_cancer_ridge_confounders"] = (
        data["Crude Rate"] - oof_cancer_ridge
    )
```

The final step will be to build a model benchmark table, that will be stored in a new `csv` file: `wi_model_results_tableau.csv`

```python
model_tables = []
for df_name in ["controlled_df", "rf_df", "combined_df"]:
    if df_name in globals():
        model_tables.append(globals()[df_name].copy())

if model_tables:
    if "combined_df" in globals():
        models_out = globals()["combined_df"].copy()
    else:
        models_out = pd.concat(model_tables, ignore_index=True).drop_duplicates()

    def categorize_feature_set(name):
        if "Confounders_Only" in name:
            return "Socioeconomic Controls"
        if "All_Env" in name:
            return "All Environmental + Controls"
        if "Air" in name:
            return "Air + Controls"
        if "Water" in name:
            return "Water + Controls"
        if "Ag_Land" in name:
            return "Ag / Land + Controls"
        return "Other"

    models_out["feature_group"] = models_out["feature_set"].astype(str).apply(
        categorize_feature_set
    )
    models_out_path = "wi_model_results_tableau-copy.csv"
    models_out.to_csv(models_out_path, index=False)
    print(f"Saved {models_out_path} -> shape {models_out.shape}")
else:
    print(
        "No model results DataFrame (`controlled_df`, `rf_df`, or `combined_df`) "
        "found in memory; skipped `wi_model_results_tableau.csv`."
    )
```
